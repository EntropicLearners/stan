```python
import os
import json

json_path = "../stan/data/bindex_tab.json"
```

```python
if os.path.exists(json_path):
    print(f"Loading cached index from {json_path}...")
    with open(json_path, "r") as f:
        index_data = json.load(f)
else:
    print("Cached index not found. Please create JSON index first.")
    exit(1)

print(f"Index loaded with {len(index_data)} terms.")
```

```python
index_data[:10] # first 10 entries
```

Search high-level topics

```python
query = "entropy"
query = query.lower()

for topic in index_data:
    if topic["topic"].lower() == query:
        print(f"pages: {topic["pages"]}")
        print(topic["subtopics"])
```

Notice the use of the get method here!

## Ideas
Passing the JSON format directly to the LLM might work very well... here is an example I tested manually with command-r7b. The CONTEXT was produced with


```python
for topic in index_data:
    if query in topic["topic"].lower():
        print(topic)
        print(f"pages: {topic.get('pages', '(no pages)')}")
        print(f"sub-topics: {topic.get('subtopics', '')}")
```
