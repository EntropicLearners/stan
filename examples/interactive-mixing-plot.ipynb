```python
import numpy
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def mixing_energy(x1, G1, G2, A):
    # Linear mixing term
    linear_mixing = x1 * G1 + (1 - x1) * G2
    
    # Entropy term
    entropy = numpy.where((x1 > 0) & (x1 < 1), x1*numpy.log(x1) + (1-x1)*numpy.log(1-x1), 0)
    
    # Ideal mixing
    ideal_mixing = linear_mixing + entropy

    # Full expression with A term
    full_energy = ideal_mixing + A * x1 * (1 - x1) 
    
    return full_energy, ideal_mixing, linear_mixing

def plot_mixing_energy(A):

    # Setup parameters
    G1 = 0.2
    G2 = 0.1

    # Create x1 array
    x1 = numpy.linspace(0, 1, 200)
    
    # Calculate energies
    full_energy, ideal_mixing, linear_mixing = mixing_energy(x1, G1, G2, A)
    
    # Create plot
    plt.figure(figsize=(10, 6))
    plt.plot(x1, full_energy, label=f'Full Energy (A={A})', color='red')
    plt.plot(x1, linear_mixing, label='Linear Mixing', color='blue', linestyle='--')
    plt.plot(x1, ideal_mixing, label='Ideal Mixing', color='green', linestyle='--')

    plt.title('Mixing Energy Landscape')
    plt.xlabel('x1')
    plt.ylabel('Energy')
    plt.legend()
    plt.grid(True)
    plt.show()

# Create widgets
A_slider = widgets.FloatSlider(
    value=0.0,
    min=-2.0,
    max=5.0,
    step=0.1,
    description='A:',
    continuous_update=False
)

# Interactive plot function
interactive_plot = widgets.interactive(
    plot_mixing_energy,
    A=A_slider
)

# Display widgets
display(interactive_plot)
```
