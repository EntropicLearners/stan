```python
import numpy

def calculate_henrys_law_problem (x, h=990, p2vap=0.01227):
    p_total = (x*h) + (1-x)*p2vap
    x1 = p_total/h
    y2 = (1-x1)*p2vap/p_total
    y1 = 1-y2
    print(f'P_total = {p_total}')
    print(f'x1      = {x1}')
    print(f'y2      = {y2}')
    print(f'y1      = {y1}')
    print(f'y1 + y2 = {y1+y2}')

calculate_henrys_law_problem(0.01)
```
