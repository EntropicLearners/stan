```python
import matplotlib.pyplot as plt
import numpy as np

# Create the figure and axis
plt.figure(figsize=(6, 6))
ax = plt.gca()

# Plot the diagonal line with a valid dark green color
plt.plot([-1, 1], [-1, 1], color='#006400', linewidth=2)  # Using hex code for dark green

# Set the axis limits
plt.xlim(-1, 1)
plt.ylim(-1, 1)

# Remove axes labels and ticks
plt.xlabel('')
plt.ylabel('')
ax.set_xticklabels([])
ax.set_yticklabels([])
ax.set_xticks([])
ax.set_yticks([])

# Position the spines at the center (0,0)
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')

# Ensure the aspect ratio is equal
plt.axis('equal')

# Show the plot
plt.tight_layout()
plt.show()
```
